```python
oscar_movies=[
    "dark knight",
    "hurt locver",
    "hi",
    "hello"
]

years=[2008,2009,2010,2012]
oscar_revenue=[1086,974,333,345]
plt.title("Oscar movies revenue each year")
plt.xlabel("years")
plt.ylabel("revenue")

plt.plot(years,oscar_revenue)
```

```python
non_oscar_movies=[
    "dark knight1",
    "hurt locver1",
    "hi1",
    "hello1"
]

years=[2008,2009,2010,2012]
non_oscar_revenue=[108,1724,1333,1345]
plt.plot(years,oscar_revenue,"o--r",linewidth=5,label="oscar movies")
plt.plot(years,non_oscar_revenue,"<:b",label="non oscar movies")

plt.title("Oscar movies vs Non oscar")
plt.xlabel("years")
plt.ylabel("revenue")
plt.style.available
plt.style.use("default")
plt.grid()
plt.legend()
plt.savefig("finalplot.png")
```

```python
movies=[
    "dark knight1",
    "hurt locver1",
    "hi1",
    "hello1"
]

years=[2008,2009,2010,2012]
revenue=[108,1724,1333,1345]
plt.plot(movies,revenue)
plt.tight_layout()
```

```python
# show 2 on bar chart
import numpy as np

oscar_movies=[
    "dark knight",
    "hurt locver",
    "hi",
    "hello"
]

years=[2008,2009,2010,2012]
oscar_revenue=[1086,974,333,345]
non_oscar_movies=[
    "dark knight1",
    "hurt locver1",
    "hi1",
    "hello1"
]
x=np.arange(len(years))
years1=[2008,2009,2010,2012]
non_oscar_revenue=[108,1724,1333,1345]
width=0.4
plt.bar(x-width/2,oscar_revenue,width,color="orange")
plt.bar(x+width/2,non_oscar_revenue,width,color="brown")
plt.xticks(x,years)
```

```python
# horizontal chart
oscar_movies=[
    "dark knight",
    "hurt locver",
    "hi",
    "hello"
]

years=[2008,2009,2010,2012]
oscar_revenue=[1086,974,333,345]
plt.barh(oscar_movies,oscar_revenue)
```

```python
for i in range(len(years)):
    plt.annotate(oscar_movies[i],xy=(years[i],oscar_revenue[i]),xytext=(years[i],oscar_revenue[i]))
plt.scatter(years,oscar_revenue,color="brown")
plt.grid()
```

```python

plt.pie(oscar_revenue,labels=oscar_movies,autopct="%1.1f%%",
        wedgeprops={
        "edgecolor":"black"
        }
       )
```
