```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Data preprocessing
# handle missing values
```

```python
df["Income"]=df["Income"].fillna(df["Income"].median())
```

```python
# Feature Engineering
# convert year of birth to age
```

```python
df["Age"]=2026-df["Year_Birth"]
df.head()
```

```python
# date of joining to customer tenure
```

```python
df["Dt_Customer"]=pd.to_datetime(df["Dt_Customer"],dayfirst=True)
reference_date=df["Dt_Customer"].max()
print(reference_date)
```

```python
df["Customer_Tenure_Days"]=(reference_date-df["Dt_Customer"]).dt.days
```

```python
df.head()
```

```python
#  Total spending
df["Total_Spending"]=df["MntWines"]+df["MntFruits"]+df["MntMeatProducts"]+df["MntFishProducts"]+df["MntSweetProducts"]+df["MntGoldProds"]
df["Total_Spending"]
```

```python
 # children
df["Total_Children"]=df["Kidhome"]+df["Teenhome"]
df["Total_Children"]
```

```python
# Education
df["Education"].value_counts()
```

```python
# Undergra,grad,postgrad
```

```python
df["Education"]=df["Education"].replace({"Graduation":"Graduate","Basic":"Undergraduate","PhD":"Postgraduate","Master":"Postgraduate","2n Cycle":"Undergraduate"})

df["Education"]                 
```

```python
df["Education"].value_counts()
```

```python
# Marital status
df["Marital_Status"].value_counts()
```

```python
df["Living_With"]=df["Marital_Status"].replace({"Married":"Partner","Together":"Partner","Single":"Alone","Divorced":"Alone","Widow":"Alone","Absurd":"Alone","YOLO":"Alone"})

df["Living_With"]
```

```python
df["Living_With"].value_counts()
```

```python
df.head()
```

```python
# Encoding
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe=OneHotEncoder()
cat_cols=["Education","Living_With"]
enc_cols=ohe.fit_transform(df_cleaned[cat_cols])
```

```python
enc_df=pd.DataFrame(enc_cols.toarray(),columns=ohe.get_feature_names_out(cat_cols),index=df_cleaned.index)
```

```python
enc_df.head()
```

```python
df_encoded=pd.concat([df_cleaned.drop(columns=cat_cols),enc_df],axis=1)
```

```python
df_encoded.head()
```

```python
# Scaling
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
X=df_encoded

```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
# visualising
```

```python
X_scaled.shape
```

```python
# 2d visualisation
from sklearn.decomposition import PCA 
```

```python
pca=PCA(n_components=3)
X_pca=pca.fit_transform(X_scaled)
```

```python
# plot
fig=plt.figure(figsize=(8,6))
ax=fig.add_subplot(111, projection="3d")
plt.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2])
ax.set_xlabel("PCA1")
ax.set_ylabel("PCA2")
ax.set_zlabel("PCA3")
```

```python
pca.explained_variance_ratio_
```

```python
# identify k, use elbow method and silhoutte scoring for that
```

```python
from sklearn.cluster import KMeans
from kneed import KneeLocator
wcss=[]
for k in range(1,11):
    kmeans=KMeans(n_clusters=k,random_state=42)
    kmeans.fit_predict(X_pca)
    wcss.append(kmeans.inertia_)
    
```

```python
plt.plot(range(1,11),wcss,marker='o')
```

```python
from kneed import KneeLocator
knee=KneeLocator(range(1,11),wcss,curve="convex",direction="decreasing")
```

```python
knee.elbow
```

```python
# best k is 4 yet
```

```python
# silhoutte score
```

```python
from sklearn.metrics import silhouette_score
```

```python
ss=[]
for k in range(2,11):
    kmeans=KMeans(n_clusters=k,random_state=42)
    labels=kmeans.fit_predict(X_pca)
    score=silhouette_score(X_pca,labels)
    ss.append(score)
```

```python
plt.plot(range(2,11),ss,marker='o')
```

```python
# combined plot
k_range=range(2,11)
fig,ax1=plt.subplots(figsize=(8,6))
ax1.plot(k_range,wcss[:len(k_range)],marker="o",color="blue")
ax1.set_xlabel("K")
ax1.set_ylabel("WCSS")

ax2=ax1.twinx()
ax2.plot(k_range,ss[:len(k_range)],marker="x",color="red",linestyle='--')
```

```python
# choose nearest intersection k
```

```python
# Clustering
# Kmeans
```

```python
kmeans=KMeans(n_clusters=4,random_state=42)
labels_kmeans=kmeans.fit_predict(X_pca)
```

```python
# plot
fig=plt.figure(figsize=(8,6))
ax=fig.add_subplot(111, projection="3d")
plt.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2],c=labels_kmeans)
ax.set_xlabel("PCA1")
ax.set_ylabel("PCA2")
ax.set_zlabel("PCA3")
```

```python
# Aglomorative clustering
from sklearn.cluster import AgglomerativeClustering
```

```python
agg_clf=AgglomerativeClustering(n_clusters=4,linkage="ward")
labels_agg=agg_clf.fit_predict(X_pca)
```

```python
fig=plt.figure(figsize=(8,6))
ax=fig.add_subplot(111, projection="3d")
plt.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2],c=labels_agg)
ax.set_xlabel("PCA1")
ax.set_ylabel("PCA2")
ax.set_zlabel("PCA3")
```
