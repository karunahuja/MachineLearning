```python
from sklearn.svm import SVR
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import datasets
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score


df = datasets.load_diabetes(as_frame=True).frame
df.head()
```

```python
X=df.drop("target",axis=1)
```

```python
y=df["target"]
```

```python
X.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)
```

```python
y
```

```python
y_scaler=StandardScaler()
y_train_scaled=y_scaler.fit_transform(y_train.values.reshape(-1,1)).ravel()
y_test_scaled=y_scaler.transform(y_test.values.reshape(-1,1)).ravel()
```

```python
y_train_scaled
```

```python
y_test_scaled
```

```python
model=SVR()
model.fit(X_train,y_train_scaled)
```

```python
y_pred_scaled=model.predict(X_test)
```

```python
r2_score(y_test_scaled,y_pred_scaled)
```

```python
# Linear model
model=SVR(kernel="linear")
model.fit(X_train,y_train_scaled)
y_pred_scaled=model.predict(X_test)
r2_score(y_test_scaled,y_pred_scaled)
```

```python
model=SVR(kernel="sigmoid")
model.fit(X_train,y_train_scaled)
y_pred_scaled=model.predict(X_test)
r2_score(y_test_scaled,y_pred_scaled)
```

```python
model=SVR(kernel="poly")
model.fit(X_train,y_train_scaled)
y_pred_scaled=model.predict(X_test)
r2_score(y_test_scaled,y_pred_scaled)
```

```python
# overfitiing is happening lets do next step
# Hyperparameter tuning with gridsearch cv

from sklearn.model_selection import GridSearchCV




param_grid = {
    'kernel': ['linear', 'rbf'],
    'C': [1,2,5,10,50,100],
    'epsilon': [0.01,0.1,0.2,0.3,0.5]
}



svr=SVR()

grid_search = GridSearchCV(svr, param_grid, cv=3,scoring="r2")
grid_search.fit(X_train, y_train_scaled)

print(grid_search.best_params_)
```

```python
best_model=SVR(kernel="linear",C=10,epsilon=0.01)
best_model.fit(X_train,y_train_scaled)
y_pred_scaled=best_model.predict(X_test)
y_train_pred_scaled=best_model.predict(X_train)
r2_score(y_test_scaled,y_pred_scaled)
```

```python
r2_score(y_train_scaled,y_train_pred_scaled)
```

```python
# here train and test r2 score is close so no overfitting hence best case
```
