```python
import json
```

```python
# Load the data
def load_data(filename):
    with open(filename,"r") as f:
        data=json.load(f)
    return data
```

```python
data=load_data("store_data.json")
print(data)
```

```python
cleaned_data=[]
def clean_data(data):
    text_to_num={"one":1,"two":2,"three":3,"four":4}

    unique_users=set()
    for user in data:
        raw_rating=user["rating"]
        if(raw_rating in text_to_num):
            raw_rating=text_to_num[raw_rating]
        user["rating"]=raw_rating
        # handle missing values
        raw_age=user.get("age")
        if(raw_age==None):
            user["age"]=None
        #deduplication
        if(user["name"].strip() in unique_users):
            continue   
        unique_users.add(user["name"])
        cleaned_data.append(user)
        
    print(cleaned_data)
```

```python
clean_data(data)
```

```python
# Avg rating and poor rating
def get_insights(data):
    # avg rating
    tot_rating=0
    count=0
    for user in data:
        tot_rating+=user["rating"]
        count+=1
    avg_rating=tot_rating/count
    print(avg_rating)

    # poor ratings
    poor_ratings=0
    for user in data:
        if(float(user["rating"])<3):
            poor_ratings+=1
    print(f"{poor_ratings/len(data)*100}")
get_insights(cleaned_data)
```

```python
# Build recommendation feature

def get_recommendations(data):
    recommendations=[]
    for user in data:    
        curr_recomm={}
        curr_recomm["name"]=user["name"]
        if(float(user["rating"])>=4):
            curr_recomm["brand"]="apple"
        else:
            curr_recomm["brand"]="samsung"
        recommendations.append(curr_recomm)
    return recommendations
```

```python
print(get_recommendations(cleaned_data))
```
