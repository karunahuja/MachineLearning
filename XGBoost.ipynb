```python
import xgboost as xgb
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,classification_report

```

```python

# 1. Generate a synthetic dataset
X, y = make_classification(n_samples=500, n_features=20, random_state=42,n_informative=10,n_redundant=2)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 2. Initialize the XGBoost Classifier using standard hyperparameters
model = xgb.XGBClassifier(
    n_estimators=100,       # Number of gradient boosted trees
    max_depth=3,            # Maximum tree depth for base learners
    learning_rate=0.1,      # Boosting learning rate (eta)
    # Loss function to optimize
    random_state=42,
    eval_metric="logloss"
)

# 3. Fit the model on training data
model.fit(X_train, y_train)

# 4. Predict and evaluate
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Scikit-Learn API Accuracy: {accuracy * 100:.2f}%")
```

```python
classification_report(y_test,y_pred)
```
